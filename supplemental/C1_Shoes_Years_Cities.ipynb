## Q4 · Year eras with pd.cut
> **Quiz Q4:** Run the **Q4** cell. It turns `year` into numbers and sorts them into four eras with `pd.cut`. How many answers are numbers, out of how many? Give the four era counts.Bins in `pd.cut` include their right edge but not their left edge. Use that to say which era an answer of exactly `1900` gets, and which era the 20 answers of exactly `3000` get.
```python
yn = pd.to_numeric(year, errors='coerce')
print("numbers:", yn.notna().sum(), "of", len(year))

era = pd.cut(yn, bins=[0, 1900, 2026, 3000, 1000000],
             labels=['long ago', 'recent past', 'near future', 'far future'])
print(era.value_counts(sort=False))
print("a number but no era:", (yn.notna() & era.isna()).sum(), yn[yn.notna() & era.isna()].tolist())
```

## Q8 · Pull out the digits (extra credit)
> **Quiz Q8:** **First, before running anything, predict:** about 30 shoe answers are still not numbers ("5?", "3 or 4", "2 pairs", ...). If you rescue them, will the mean of about 3.5 pairs move much?Run the **Q8** cell. It adds a rescue step: `.str.extract(r'(\d+)')` pulls the first run of digits out of each answer. What does it give for `"3 or 4"` and `"10-12"`? How many numbers now, and what is the mean?Add a line `fixed.sort_values(ascending=False).head(3)` and run it. Which answer is responsible for the jump? Choose a rule for believable answers and give the count and mean under your rule.

**Tell your Recorder:** your prediction; the values for "3 or 4" and "10-12", the new count and mean; then the answer responsible, your rule, and the count and mean under it.

```python
n = pd.to_numeric(shoes, errors='coerce')
words = {'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5}
fixed = n.fillna(shoes.str.strip().str.lower().map(words))
digits = pd.to_numeric(shoes.str.extract(r'(\d+)')[0])
fixed = fixed.fillna(digits)
print("numbers:", fixed.notna().sum(), "of", len(shoes), "| mean:", round(fixed.mean(), 2))
```
