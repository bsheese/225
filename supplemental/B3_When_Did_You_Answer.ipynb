## Q7 · How long did a class take?
> **Quiz Q7:** Run the **Q7** cell. For each day in 2026 it counts the responses and finds the first and last time, then shows the three busiest days. Read the code:What does `t26.dt.date` turn each time stamp into? (Look at the left-hand column of the table.) Why does that make `groupby` give one row per day?What is `g['span']`?The two busiest days had 19 and 14 responses. How many minutes did each take from first to last response? Which one looks like a single class period, and what could explain the other?

**Tell your Recorder:** what t26.dt.date and span are, the minutes for both busiest days, then your interpretation of each.

```python
t26 = ts[ts.dt.year == 2026]
g = t26.groupby(t26.dt.date).agg(['count', 'min', 'max'])
g['span'] = g['max'] - g['min']
g['minutes'] = g['span'].dt.total_seconds() / 60
g.sort_values('count', ascending=False).head(3)
```
