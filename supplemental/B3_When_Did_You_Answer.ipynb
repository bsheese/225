## Q8 · How long did a class take?
> **Quiz Q8:** Run the **Q8** cell. For each day in 2026 it shows how many responses came in and the first and last time. Read the code: what does `ts.dt.date` do, and what is `g['span']`? Add this line at the end of the cell and run it again:
```
g["minutes"] = g["span"].dt.total_seconds() / 60
g.sort_values("count", ascending=False).head(3)
```
The two busiest days had 19 and 14 responses. How many minutes did each take from first to last response? Which one looks like a single class period, and what could explain the other?

**Tell your Recorder:** what ts.dt.date and span are, the minutes for the two busiest days, then your interpretation.

````python
t26 = ts[ts.dt.year == 2026]
g = t26.groupby(t26.dt.date).agg(['count', 'min', 'max'])
g['span'] = g['max'] - g['min']
g.sort_values('count', ascending=False).head(3)
````

## Q10 · Gaps between responses
> **Quiz Q10:** Run the **Q10** cell. What is the median gap in minutes, how many gaps are longer than 30 days, and what is the longest gap? What do the long gaps represent?Now a classmate's version: in the **last line** of the cell, change `.dt.total_seconds()` to `.dt.seconds` and run it. What does it now say the longest gap is, in hours? Explain the difference between `.dt.seconds` and `.dt.total_seconds()`.

**Tell your Recorder:** the median gap, the count over 30 days, the longest gap, what the long gaps are, then the classmate's number and your explanation.

````python
gaps = ts.sort_values().diff()
print("median gap (minutes):", gaps.dt.total_seconds().median() / 60)
print("gaps over 30 days:", (gaps > pd.Timedelta(days=30)).sum())
print("longest gap:", gaps.max())
print("longest gap in hours:", gaps.dt.total_seconds().max() / 3600)
````
